```python
import numpy as np
import pandas as pd

# para leer el .data, al no ser csv(comma-separated values) se necesita adaptar
df = pd.read_csv('../../datos/housing.data', header =None, sep='\s+')

df.columns= ['CRIM','ZN','INDUS','CHAS','NOX','RM','AGE','DIS','RAD','TAX','PTRATIO','B','LSTAT','MEDV']

df.head()
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt

sns.set (style= 'whitegrid', context='notebook')
cols = ['DIS', 'INDUS', 'CRIM', 'RM', 'MEDV']
sns.pairplot(df[cols], height = 2.5) 
plt.show()

# se debe usar un metodo numerico para poder entender las correlaciones
```

```python
# como se realcionan las variables (promedio de las habitaciones que existen en las casas)
# se va a hacer un mapa de calor
import numpy as np 
# se agrega una matriz transpuesta
cm = np.corrcoef(df[cols].values.T)
sns.set(font_scale = 1.5)
# cbr -> barra al lado de la info, 
sns.heatmap(cm, cbar=True, annot=True, yticklabels=cols, xticklabels=cols)
```

```python
# K-NN
# “Dime quiénes son tus vecinos más cercanos y te diré quién eres.”
# Se usa para clasificacción y regresión.
# No aprende un modelo, solo almacena los datos de entrenamiento, elige un valor de K (numero de vecinos)-> calcula la distancia entre el nuevo dato y los datos de entrenamiento -> elige los K vecinos más cercanos -> toma una decisión basada en esos vecinos (votación para clasificación, promedio para regresión).

# K -> Número de vecinos a considerar, para medir la distancia -> Euclidiana, Manhattan, Minkowski, Hamming, coseno.
# SI K es muy pequeño -> modelo sensible al ruido.
# SI K es muy grande -> modelo demasiado general (underfitting).

# Ventajas== facil de entender e implementar, no requiere entrenamiento, puede manejar datos no lineales, funciona bien con conjuntos de datos pequeños.
# Desventajas== computacionalmente costoso con grandes conjuntos de datos, sensible a ruido y datos irrelevantes, requiere escalado de características, no maneja bien datos dispersos.

# Necista de escalado de datos(normalización o estandarización) para que las distancias sean significativas.
```

```python
# SVR (Support Vector Regression) -> Suport Vector Machines aplicado a regresión
# sirve para predecir valores numericos

# “Los errores pequeños no me importan, solo los grandes”
# ε (epsilon)-insensitive tube -> margen de tolerancia donde los errores no se penalizan.
# Función de pérdida -> penaliza los errores fuera del margen ε.
# Kernel trick -> transforma los datos a un espacio de mayor dimensión para manejar relaciones no lineales.
# Parámetros importantes -> C (regularización) c-underfifting C-overfifting, ε (margen de tolerancia), tipo de kernel (lineal, polinomial, RBF).
# support vectors -> puntos de datos que definen la posición del modelo.
# 
# uso de kernel si no es lineal
# Ventajas== maneja relaciones no lineales, robusto a outliers, efectivo en espacios de alta dimensión.
# Desventajas== computacionalmente costoso, sensible a la elección de hiperparámetros, no escala bien con grandes conjuntos de datos. 

import matplotlib.pyplot as plt
import matplotlib.image as npimg

# imagen  que muestra svr básico 
img = npimg.imread('../../img/svr_basic.png')
plt.imshow(img)
plt.axis('off')

# imagen que muestra svr con kernel
img2 = npimg.imread('../../img/svr_2.png')
plt.imshow(img2)
plt.axis('off')
```

```python
# Teoria
# ¿Qué es Random Forest?
    # # Random Forest es un conjunto de muchos árboles de decisión que trabajan juntos para tomar una decisión más estable y confiable.

# ¿Qué modelo base usa?
    # # Usa árboles de decisión como modelos base.

# ¿Clasificación, regresión o ambos?
    # # RandomForestClassifier
    # Se usa cuando la variable objetivo es categórica, cada arbol vota por una clase la clase con mas votos es la predicción final(Accuracy--Precision / Recall--F1-score--ROC-AUC)

    # RandomForestRegressor
    # Se usa cuando la variable objetivo es numérica continua, Cada árbol predice un valor numérico y la predicción final es el promedio de todas las predicciones de los árboles(MAE--MSE--RMSE--R²)

# ¿Qué significa que Random Forest sea un método de ensamble?
    # Un método de ensamble combina varios modelos para obtener una predicción mejor que la de un solo modelo.

# ¿Por qué Random Forest usa muchos árboles en lugar de uno solo?
    # Porque un solo árbol se sobreajusta fácilmente, pero muchos árboles promedian errores y hacen el modelo más robusto.

# ¿Qué es el bootstrapping y para qué se usa en Random Forest?
    # se crean muchos arboles en conjunto, cada arbol se entrena con una muestra aleatoria de los datos (con reemplazo) para aumentar la diversidad entre los árboles y reducir el sobreajuste.

# ¿Qué ocurre en la predicción final en: a) clasificación b) regresión
    # CLASIFICACIÓN ->Cada árbol vota → gana la clase más votada
    # REGRESIÓN ->Cada árbol predice un valor → se promedian las predicciones

# ¿Qué diferencia clave hay entre un árbol de decisión normal y uno dentro de un Random Forest?
# ARBOL DE DECISIÓN NORMAL -> se entrena con todos los datos y todas las características
# ARBOL EN RANDOM FOREST -> se entrena con una muestra aleatoria de los datos y un subconjunto aleatorio de características

# ¿Qué controla el parámetro n_estimators?
    # Controla el número de árboles en el bosque. Más árboles suelen mejorar la precisión hasta cierto punto, pero aumentan el tiempo de entrenamiento

# ¿Para qué sirve max_depth y qué pasa si es muy grande?
    # Profundidad grande → modelo muy complejo

# ¿Qué hace max_features y por qué ayuda a que los árboles sean “aleatorios”?
    # Controla cuántas características se consideran para cada división. Menos características → más diversidad entre árboles

# ¿Qué problema ayuda a reducir Random Forest: overfitting o underfitting? ¿Por qué?
    # Random Forest reduce el overfitting porque promedia modelos complejos (árboles) para obtener una predicción más estable.

# ¿Qué es la importancia de variables (feature importance) en Random Forest?
    # Es una medida de qué tan importante es cada variable para tomar decisiones en el modelo.

# ¿Por qué puede ser engañosa la feature importance tradicional?
    # por que favorece variables aunque no sean relevantes, se usa SHAP o Permutation Importance (explica el por qué de las predicciones)

# Menciona una métrica para evaluar Random Forest en:# a) clasificación # b) regresión
    # CLASIFICACION 
    # Acuarcy -> proporción de predicciones correctas
    # Precision -> proporción de verdaderos positivos entre todos los positivos predichos (De los que predije como “Sí”, cuántos eran realmente “Sí”?)
    # Recall -> proporción de verdaderos positivos entre todos los positivos reales (De los “Sí” reales, cuántos detecté?)
    # F1-score -> media armónica de precisión y recall (balance entre precisión y recall) si una es baja, el F1 será bajo
    # ROC-AUC -> mide la capacidad del modelo para distinguir entre clases (Qué tan bien separa el modelo las clases)

# ¿En qué casos NO sería buena idea usar Random Forest?

    # Necesitas interpretabilidad total, Tienes datos muy grandes y poco cómputo, Los datos son muy escasos, se requiere un modelo muy rápido en predicción

# ¿Random Forest necesita normalizar o escalar los datos? ¿Por qué?
    # No, porque los árboles de decisión no se ven afectados por la escala de las características.

# Si tu modelo tiene muy buen resultado en entrenamiento pero malo en test, ¿qué hiperparámetro podrías ajustar para mejorar esto?
    # Disminuir max_depth --- ajustar min_samples_leaf --- Aumentar min_samples_split --- aumentar n_estimators

# CASTIGO == REGULARIZACION
```

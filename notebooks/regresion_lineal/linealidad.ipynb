```python
# ¿Cuándo usar regresión lineal?
# # Uso regresión lineal cuando la variable objetivo es numérica y la relación es aproximadamente lineal.

# ¿Cuándo NO usar regresión lineal?
# # cuando se tratan de datos cualitativos, No usar cuando hay ,imágenes, texto o datos categóricos sin procesar.PERO Sí puede usarse si conviertes los datos a números

# ¿Qué métrica usarías?
# # MAE → fácil de interpretar
# # MSE → penaliza errores grandes
# # R² → entender varianza explicada ;; Puede ser alto y el modelo ser malo,  No dice cuánto te equivocas

# ¿Por qué usar Ridge o Lasso?
# # "para que no ocurra overfitting, es decir que el modelo esté muy complejo”, Reduce pesos, Evita que una variable domine, Maneja colinealidad (problemas entre variables, estan fuertemente relacionadas)
```

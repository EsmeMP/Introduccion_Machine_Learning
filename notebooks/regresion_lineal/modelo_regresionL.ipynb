```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# para leer el .data, al no ser csv(comma-separated values) se necesita adaptar
df = pd.read_csv('../../datos/housing.data', header =None, sep='\s+')

df.columns= ['CRIM','ZN','INDUS','CHAS','NOX','RM','AGE','DIS','RAD','TAX','PTRATIO','B','LSTAT','MEDV']

df.head()
```

```python
# # modelo de regresion lineal -> utiliza puntos reales, se ajusta a una linea recta para hacer predicciones 

# crea el modelo con sklearn
# libreria de ML -> REGRESION LINEAL
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

x = df['RM'].values.reshape(-1,1)
y = df['MEDV'].values.reshape(-1,1)

sc_x = StandardScaler()
sc_y = StandardScaler()

#  ajustar los datos
x_std = sc_x.fit_transform(x)
y_std = sc_y.fit_transform(y)

slr = LinearRegression()
slr.fit(x_std, y_std)
```

```python
# analisis de resultados
plt.scatter(x_std, y_std)
# muestra la linea
plt.plot(x_std, slr.predict(x_std), color = 'navy')
plt.ylabel('Mediana del precio de las casas en miles [MEDV]')
plt.xlabel('Promedio de habitaciones [RM]')
```

```python
# empezar a hacer predicciones
num_habitaciones = 5
num_habitaciones_std = sc_x.transform(np.array(num_habitaciones).reshape(-1,1))
print('Precio de una casa con 5 habitaciones en boston es de: ', sc_y.inverse_transform(slr.predict(num_habitaciones_std)))
slr.predict(num_habitaciones_std)
```

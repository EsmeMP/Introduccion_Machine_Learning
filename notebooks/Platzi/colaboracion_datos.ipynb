```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# cargar datos partidos
df_partidos = pd.read_csv('../datos/partidos_cebollitas_fe82a1a4-e109-41b1-8b78-d9b4341dacaf.csv')

# mostrar primeras filas
df_partidos.head(6)
```

```python
# estadisticas descriptivas generales
df_partidos.describe()
```

```python
# promedio de goles de cebollitas como local y como visitantes
goles_local_cebo = df_partidos[df_partidos['equipo_local'] == 'Cebollitas FC']['goles_local'].mean()
goles_visitante_cebo = df_partidos[df_partidos['equipo_visitante'] == 'Cebollitas FC']['goles_visitante'].mean()

print('goles local de cebollitas->',goles_local_cebo)
print('goles visitante de cebollitas->', goles_visitante_cebo)
```

```python
# histogramas de goles (local & visitante)
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df_partidos['goles_local'], kde=True, ax=ax[0], bins=10, color='deeppink')
ax[0].set_title('Distribucion de goles del equipo local')

sns.histplot(df_partidos['goles_visitante'], kde=True, ax=ax[1], bins=10, color='darkslategray')
ax[1].set_title('Distribucion de goles del equipo visitante')

plt.show()
```

```python
# boxplot para goles de equipo local
sns.boxplot(x=df_partidos['goles_local'], color='honeydew')
plt.title('Boxplot de goles local en cebollitas :3')
plt.show
```

```python
# scatterplot de posesion vs goles marcados(local)
sns.scatterplot(x='posesion_local (%)', y='goles_local', data=df_partidos, color='purple')
plt.title('Relacion posicion vs goles marcados (equipo local)')
plt.show()
```

```python
# mapa de calor para correlacion entre variables clave
plt.figure(figsize = (10,6))
sns.heatmap(df_partidos[['goles_local', 'goles_visitante', 'posesion_local (%)', 'posesion_visitante (%)', 'tiros_arco_local', 'tiros_arco_visitante']].corr(), annot=True, cmap='coolwarm')
plt.title('Mapa de calor -> correlacion de variables')
plt.show()
```

```python
# COMO AFECTA LA REGULARIZACION SIN Y CON L2(REDUCE PESOS GRANDES)
import numpy as np
import matplotlib.pyplot as plt

# Datos de ejemplo
x = 3
y_true = 10

# Rango de posibles valores para w
w = np.linspace(-10, 10, 500)

# Pérdida sin regularización
loss_no_reg = (y_true - w * x)**2

# Pérdida con regularización L2
lambda_reg = 5
loss_reg = (y_true - w * x)**2 + lambda_reg * (w**2)

plt.figure(figsize=(8, 5))
plt.plot(w, loss_no_reg, label="Sin regularización")
plt.plot(w, loss_reg, label=f"Con L2 (λ = {lambda_reg})")
plt.axvline(0, color='gray', linestyle='--')
plt.xlabel("Peso w")
plt.ylabel("Pérdida")
plt.title("Comparación visual: pérdida con y sin L2")
plt.legend()
plt.grid()
plt.show()
```

```python
#  L2 afecta el entrenamiento de un modelo
import numpy as np

# Peso inicial
w = 8.0       # un peso grande adrede
x = 3
y_true = 10
lr = 0.01
lambda_reg = 0.5

print("Entrenamiento con regularización L2 (gradiente descendente):\n")

for epoch in range(1, 31):
    # Predicción
    y_pred = w * x
    
    # Pérdida con L2
    loss = (y_true - y_pred)**2 + lambda_reg * w**2
    
    # Derivada (gradiente)
    dloss_dw = -2*x*(y_true - y_pred) + 2*lambda_reg*w
    
    # Actualización del peso
    w = w - lr * dloss_dw
    
    print(f"Epoch {epoch:2d} | Peso w = {w:.4f} | Pérdida = {loss:.4f}")
```

```python
# λ pequeño	
# Modelo flexible	
# Más riesgo de overfitting(sobreajuste)	
# Pesos pueden crecer	
# Puede aprender demasiado	

# λ grande
# Modelo rígido
# Más riesgo de underfitting (subajuste)
# Pesos muy pequeños
# Puede aprender muy poco
import numpy as np
import matplotlib.pyplot as plt

# Datos simples
x = 3
y_true = 10

# Rango de pesos
w = np.linspace(-10, 10, 500)

# Valores de lambda
lambdas = [0, 1, 5, 10, 20]

plt.figure(figsize=(8, 6))

for lam in lambdas:
    loss = (y_true - w * x)**2 + lam * (w**2)
    plt.plot(w, loss, label=f"λ = {lam}")

plt.xlabel("Peso w")
plt.ylabel("Pérdida")
plt.title("Regularización L2 — Comparación Visual")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd 
# divide los datos
from sklearn.model_selection import train_test_split
# controles interativos o dinamicos
import ipywidgets as widgets
from IPython.display import display

# cargar dataset original
df_partidos = pd.read_csv('../datos/partidos_cebollitas_fe82a1a4-e109-41b1-8b78-d9b4341dacaf.csv')

df_partidos['diferencia_goles'] = df_partidos['goles_local'] - df_partidos['goles_visitante']

df_partidos.head()
```

```python
# slider interactiva para elegir el porcentaje de prueba
slider_test_size = widgets.FloatSlider(
    value=0.2, min=0.1, max=0.5, step=0.05,
    description='% Test Set:', continuous_update=False
)
display(slider_test_size)
```

```python
# funcion para dividir los datos
def dividir_datos(test_size):
    x = df_partidos[['posesion_local (%)', 'tiros_arco_local', 'goles_local']]
    y = df_partidos['diferencia_goles']

    # va a usar el valor del slide
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=test_size, random_state=42)

    print(f"tamaño conjunto de entrenamiento: {len(x_train)} partidos")
    print(f"tamaño conjunto de prueba: {len(x_test)} partidos")

    return x_train, x_test, y_train, y_test

# se usa el widget interativo
widgets.interactive(dividir_datos, test_size=slider_test_size)
```

```python
# division estandar recomendada (80% entrenamiento, 20% prueba)
x_train, x_test, y_train, y_test = train_test_split(
    df_partidos[['posesion_local (%)', 'tiros_arco_local', 'goles_local']],
    df_partidos['diferencia_goles'], 
    test_size=0.2, random_state=42
)

print('Conjunto de datos de entrenamiento (primeros registros)')
display(x_train.head())

print('Conjunto de datos de prueba (primeros registros)')
display(x_test.head())
```
